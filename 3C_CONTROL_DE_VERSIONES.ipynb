# Control de Versiones

Control de versiones es la práctica o técnica de registrar y gestionar los cambios que se realizan sobre uno o varios archivos (documentos, código, imágenes, planos, etc.) a lo largo del tiempo.

# Sistema de Control de Versiones (VCS)

Un Sistema de Control de Versiones (VCS, por sus siglas en inglés Version Control System) es una herramienta que registra los cambios realizados en los archivos a lo largo del tiempo, de manera que se puedan recuperar versiones anteriores, comparar diferencias y colaborar con otras personas sin perder el historial del proyecto.

**Tipos de VCS:**



*   **Local:** Guarda versiones en un único equipo. Solo una persona lo usa
*   **Centralizado:** Hay un servidor central con el historial del proyecto. Los usuarios deben conectarse para obtener/guardar cambios.
*   **Distribuido:** Cada usuario tiene una copia completa del repositorio (historial y ramas). Permite trabajar offline y luego sincronizar.



## GIT

Git es un sistema de control de versiones distribuido (DVCS) creado por Linus Torvalds ¿reconoces ese nombre?
Git registra y gestiona los cambios en los archivos de un proyecto, permitiendo trabajar de forma segura y colaborativa.


**Funciones más significativas en Git**

* Guardar el historial de cambios
* Restaurar una versión anterior (Volver atrás en el tiempo)
* Trabajar en paralelo
* Fusionar cambios
* Colaborar con otros
* Auditar el código

### Configuración inicial

Después de instalar Git, es importante hacer una configuración de la cuenta asociada.  Esto con el fin de reconcer quien hace los cambios en los archivos.

Para conocer las configuraciones actuales podemos ejecutar el comando:


```
git config --list
```

```
git config --global --list
```
Existen 3 niveles de configuración:
* **System:** para todos los usuarios git config --system
* **Global:** para el usuario actual git config --global
* **Local:** solo para el repositorio actual git config --local

Vamos a realizar 2 configuraciones esenciales a nivel global: el nombre del usuario y su correo electrónico



```
git config --global user.name "mi nombre de usuario"
git config --global user.email "mi correo"
```
Para borrar los datos de usuario y email



```
git config --global --unset-all user.name
git config --global --unset-all user.email
```

Un repositorio Git es una carpeta a la que se le ha activado el seguimiento a sus cambios (control de versiones) en este espacio Git almacena todo el historial de cambios de un proyecto, junto con su configuración, ramas, etiquetas y metadatos.



#### git init

### Áreas en Git

Git hace seguimiento a los archivos de un repositorio usando tres áreas:

1. **Working Directory (Directorio de trabajo):** Archivos reales del proyecto que pueden ser editados manualmente.

2. **Staging Area (Área de preparación):** Archivos dispuestos para confirmar sus cambios.  Es una área intermedia con instantáneas de los archivos que han cambiado y a los cuales ponemos en el área de preparación para que sus cambios sean tenidos en cuenta en la siguiente versión del proyecto.

3. **Local Repository (Repositorio local):** Los archivos con cambios confirmados que forman parte de la nueva versión del proyecto.


image.png

#### git add

El comando **git add** se usa para agregar archivos al Staging area (Área de Preparación).  



```
git add "nombre_archivo.ext"
```
Si queremos mover al Staging Area todos los archivos con modificaciones del Working Directory, podemos usar la opción **--all** o **-A**

```
git add --all
```

Si queremos mover solo los archivos nuevos y modificados del directorio actual al área de preparación, usamos


```
git add .
```



#### git restore


Si quisiéramos sacar un archivo del Staging, lo podemos hacer con el comando **restore**

```
git restore --staged nombre_archivo.ext
```

#### git commit

El comando **git commit** es muy relevante, ya que con este comando se pasan los cambios del Área de Preparación (Staging Area) al Repositorio, generando así la última versión de este.  La opción **-m** permite incluir un mensaje descriptivo al commit.  Esto es muy útil para llevar un registro de los cambios realizados en cada commit.

```
git commit -m "Mensaje asociado al commit"
```

La opción **--amend** se usa para hacer algunas modificaciones sobre commits ya realizados.  Por ejemplo, podemos adicionar más cambios al último commit haciendo un nuevo git add y luego...


```
git commit --amend --no-edit
```

o cambiar el mensaje del último commit

```
git commit --amend -m "nuevo mensaje"
```




#### git log

El comando **git log** se usa para ver el historial de commits realizados

#### git diff

El comando **git diff** muestra la diferencia entre el Directorio de Trabajo (Working Directory) y el último commit realizado.



```
git diff
```

Si lo que queremos conocer es la diferencia entre el Área de Preparación (Staging Area) y el último commit, agregamos el parámetro --staged



```
git diff --staged
```
Por último, si queremos conocer la diferencia entre dos commits

```
git diff <commit1> <commit2>
ejemplo:
git diff 73de dbd6
```

### Borradores o versiones sin finalizar

Algunas veces estamos trabajando sobre un proyecto y hemos avanzado a un punto en el que no queremos todavía hacer commit pero que quisiéramos guardar el avance que llevamos hasta ese punto.

#### git stash

El comando **git stash** permite guardar el avance hasta el punto actual en un espacio independiente de git para poder regresar a el en cualquier otro momento.  A diferencia de un commit, los cambios guardados con git stash son temporales y no afectan al repositorio.

Deseamos guardar el avance del trabajo sin hacer commit, pero dejando el Directorio de Trabajo limpio:


```
git stash
```
Guardar el stash con un mensaje:

```
git stash push -m "Mensaje de referencia para este stash"
```
Ver la lista de stashes:

```
git stash list
```
Si quiero obtener el estado del último stash
```
git stash apply
```
Si quiero obtener el estado de un stash en particular
```
git stash apply stash@{#}  ->donde # es el número del stash deseado
```


Para obtener lo que se ha guardado en el último stash eliminándolo de la lista de stashes:
```
git stash pop
```
Obtener lo que se ha guardado en un stash específico y eliminarlo de la lista de stashes:
```
git stash pop stash@{2}
```
Eliminar un stash específico:
```
git stash drop stash@{2}
```
Borrar todos los stashs:
```
git stash clear
```


### Ramas

#### git branch

Una rama es un espacio de trabajo alterno, en el cual se pueden hacer cambios sin afectar al proyecto principal, decidiendo más adelante si lo realizado en la rama se integra o no al proyecto principal.

En todo repositorio siempre estamos trabajando en una rama, si no indicamos lo contrario, esta rama es la rama principal, convencionalmente llamada **master** o **main**.  Si queremos crear una rama alterna, usamos el comando:

```
git branch <nombre_rama>
```

Si queremos listar todas las ramas:
```
git branch
```
Veremos que la rama actual sobre la que estamos trabajando aparece indicada con un asterísco (*)

Si queremos eliminar una rama existente:
```
git branch -d <nombre_rama>
```



#### git switch
Si queremos cambiar de rama:
```
git switch <nombre_rama>
```
Ahora, si lo que queremos es crear una rama y movernos a ella de inmediato, lo podemos hacer con el siguiente comando:

```
git switch -c <nombre_rama>
```

#### git merge

Después de haber trabajado en una rama, llega el momento de unir el trabajo realizado en la rama auxiliar con lo que teníamos en la rama principal (master).  Para ello, debemos ubicarnos en la rama principal y desde allí aplicar el comando **git merge <nombre_rama>**

```
git switch master     -> nos ubicamos en la rama principal
```
```
git merge auxiliar     -> llevamos a la rama principal lo realizado en la auxiliar
```
Cuando llevamos los cambios de una rama a otra (combinamos ramas/merge) todos los commits de la rama auxiliar son llevados a la rama principal.  Esto puede ser un poco saturador.  Para combinar todos los cambios hechos en la rama auxiliar en un solo commit, podemos usar la opción:

```
git merge --squash <nombre_rama>
```

**¿Problemas al combinar las ramas?**

Algunas veces cuando se combinan ramas (se hace merge) pueden surgir problemas.  Esto se pude dar debido a cambios de la misma parte del trabajo en dos ramas distintas.  En esos casos git nos informa sobre el conflicto y podemos resolver el conflicto manualmente.   Si en algún momento lo que queremos es cancelar la integración de las ramas, podemos volver el repositorio a su estado antes de intentar el merge con el comando:

```
git merge --abort
```

## GitHub

**GitHub** es una plataforma Web que aloja repositorios Git.  Permite gestionar, compartir y colaborar en proyectos.   

### Crear cuenta y repositorio remoto

Lo primero que debemos hacer para usar GitHub es crear una cuenta en: https://github.com/

Luego de creada la cuenta accedemos al espacio asignado y creamos en él un **Nuevo Repositorio**

Este nuevo repositorio es un **Repositorio Remoto** el cual es una versión de un repositorio local que es compartido, actualizado y gestionado en línea por uno o varios usuarios.



image.png

### Conectar el repositorio local con el remoto

Debemos indicarle a git con cual repositorio remoto se conectará nuestro repositorio local.  Para ello, desde GitHub copiamos la ruta del repositorio remoto recién creado:


image.png

Ahora, desde el shell de git de nuestro repositorio local ejecutamos el siguiente comando:

```
git remote add origin <ruta_del_repo_remoto>
```
ejemplo:


```
git remote add origin https://github.com/anjuvaos-sena/convocatorias-sena.git
```

Luego, podemos verificar la correcta asociación del repositorio remoto con el comando:

```
git remote -v
```



image.png

**Desvincular el repositorio**

Si queremos quitar el vínculo entre el repo local y el remoto, lo hacemos con el siguiente comando:



```
git remote remove origin
```



  ### Interacción entre el repositorio local y repositorio remoto

  Una vez tenemos una correcta asociación de un repositorio remoto, podemos realizar ciertas operaciones entre ellos.
  
   

#### git push

Con el comando **git push** Podemos llevar una copia actualizada de nuestro repositorio local hacia nuestro repositorio remoto.  
```
git push -u origin master
```
git pide los datos de autenticación de GitHub para poder ejecutar esta operación.


En este punto:
* **origin** hace referencia al repo remoto.  
* **master** es el nombre de la rama principal del repo local, también podría llamarse **main** si así está configurada.
* **-u** le indica a git que mantenga esta relación entre la rama principal del repo local (master) y la rama principal del repo remoto.  Esta opción **-u** solo es necesaria en el primer push, de allí en adelante, solo bastará con **git push** y git recordará la asociación.




#### git pull

Con git pull nos traemos la última versión del repositorio remoto hacia nuestro repositorio local (actualizamos los cambios hechos por cualquier otra persona)

```
git pull origin master
```
o, simplemente **git pull** si se usó la opción **-u** en el primer push
```
git pull
```




#### git clone

El comando **git clone** permite obtener una copia de un repositorio remoto en el equipo local.



```
git clone <ruta_del_repo_remoto>
```

Con este comando:


*   Se crea una copia local del repositorio remoto con todos los archivos actualizados
*   Se obtiene todo el historial de commits
*   Enlaza el repo local con el repo remoto para futuros push o pull
